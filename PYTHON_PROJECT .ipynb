# **Stage 2: Data Cleaning, Data Transformation and Feature Engineering**

```python
# Create a working copy
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
print(df_clean.head())
```

# Handle Duplicate

```python
# Check the duplicates
duplicate_count = int(df_clean.duplicated().sum())
print(f" duplicate rows found: {duplicate_count:,}")
```

# Handle Missing Values

```python
# Missing-value
missing_before = df_clean.isna().sum()
print("Missing Value :",missing_before )
```

```python
# Fill missing Series titles with 'Total'
for col in ['Series_title_2','Series_title_3','Series_title_4','Series_title_5']:
    df_clean[col] = df_clean[col].fillna('Total')

# Remove rows with missing Data_value
df_clean = df_clean.dropna(subset=['Data_value']).copy()
print("Shape after missing handling:", df_clean.shape)
print(df_clean.isna().sum())
```

# Standardize Column Names and Review Unwanted Columns

```python
# lowercase and replace spaces
import re

def to_lower_case(column_name):
    name = str(column_name).strip()
    name = re.sub(r"([a-z0-9])([A-Z])", r"\1_\2", name)
    name = re.sub(r"[^A-Za-z0-9]+", "_", name)
    return name.strip("_").lower()

original_columns = df_clean.columns.tolist()
df_clean.columns = [to_lower_case(col) for col in df_clean.columns]

print("Column-name changes:")
display(pd.DataFrame({
    "original_name": original_columns,
    "standardized_name": df_clean.columns.tolist()
}))
```

# Data Transformation

```python
# Normalize text categories; retain the original meaning
for col in df_clean.select_dtypes(include="object"):
    df_clean[col] = df_clean[col].astype(str).str.strip()

# Validate numerical ranges
print(f"Period: min={df_clean['period'].min()}, max={df_clean['period'].max()}")
print(f"Data_value: min={df_clean['data_value'].min():.2f}, max={df_clean['data_value'].max():.2f}")
print("Cleaned shape:", df_clean.shape)
```

# Feature Engineering

```python
# Extract Year and Quarter from Period
# Example: 2026.06 = Year 2026, Quarter June
df_clean["year"] = df_clean["period"].astype(str).str[:4].astype(int)
df_clean["quarter_month"] = df_clean["period"].astype(str).str[5:].astype(int)
df_clean["quarter_label"] = df_clean["quarter_month"].map({3:'Mar', 6:'Jun', 9:'Sep', 12:'Dec'})

# GDP Measure - Production / Expenditure / Income
conditions = [
    df_clean["group"].str.contains('production', case=False, na=False),
    df_clean["group"].str.contains('expenditure', case=False, na=False),
    df_clean["group"].str.contains('income', case=False, na=False),
]
labels = ['Production', 'Expenditure', 'Income']
df_clean["gdp_measure"] = np.select(conditions, labels, default='Other')

# Decade feature
df_clean["decade"] = (df_clean["year"] // 10 * 10).astype(str) + "s"

# Growth period
df_clean["period_type"] = np.where(df_clean["group"].str.contains('Seasonally', case=False, na=False),
                                   "Seasonally Adjusted", "Actual")

print("New features created: year, quarter_month, quarter_label, gdp_measure, decade, period_type")
print(df_clean[["period","year","quarter_label","gdp_measure","decade"]].head())
print(df_clean["gdp_measure"].value_counts())
```

# **Stage 3 : Statistical analysis and visualizations**

```python
# Descriptive statistics for the main numeric measures
numeric_summary_cols = ["period", "data_value", "magnitude", "year"]
display(df_clean[numeric_summary_cols].describe().T.round(2))
```

```python
# Headline GDP by Measure - Conversion maari Summary Function
def gdp_summary(column):
    result = df_clean.groupby(column, dropna=False)["data_value"].agg(
        records="count", total_gdp="sum", avg_gdp="mean", median_gdp="median"
    )
    return result.sort_values("total_gdp", ascending=False)

for col in ["gdp_measure", "quarter_label", "decade", "period_type"]:
    print(f"\nGDP Summary by {col}:")
    print(gdp_summary(col).round(2))
```

```python
# Final evidence table
headline = df_clean[(df_clean['series_title_1'].str.contains('production', case=False, na=False)) &
                    (df_clean['series_title_2'] == 'Total') &
                    (df_clean['group'].str.contains('Chain volume, Actual', case=False, na=False))].copy()
headline = headline.sort_values('period')

print(f"Total GDP Records: {len(df_clean):,}")
print(f"Headline GDP Records: {len(headline):,}")
print(f"GDP in 1972: {headline['data_value'].iloc[0]:,.0f} Million")
print(f"GDP in 2026: {headline['data_value'].iloc[-1]:,.0f} Million")
print(f"Growth: {headline['data_value'].iloc[-1]/headline['data_value'].iloc[0]:.1f}x")
display(headline[['period','year','data_value']].tail(10))
```

# Correlation Heatmap

```python
correlation_features = ["period", "data_value", "magnitude", "year"]
spearman_corr = df_clean[correlation_features].corr(method="spearman")

plt.figure(figsize=(10, 7))
sns.heatmap(spearman_corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True)
plt.title("Spearman Correlation Between GDP Measures")
plt.tight_layout()
plt.show()
display(spearman_corr.round(2))
```

# Pivot Table - Industry vs Year Heatmap

```python
pivot_df = df_clean[(df_clean['group'].str.contains('Chain volume, Actual', case=False, na=False)) &
                    (df_clean['series_title_1'].str.contains('production', case=False, na=False))]
top5 = pivot_df['series_title_2'].value_counts().head(5).index
filtered = pivot_df[pivot_df['series_title_2'].isin(top5)]
filtered = filtered[filtered['year'] >= 2020]

gdp_pivot = pd.pivot_table(filtered, index='series_title_2', columns='year',
                            values='data_value', aggfunc='mean', fill_value=0)

plt.figure(figsize=(12,6))
sns.heatmap(gdp_pivot, annot=True, fmt='.0f', cmap='YlOrRd')
plt.title('GDP by Industry vs Year (2020-2026)')
plt.xlabel('Year')
plt.ylabel('Industry')
plt.show()
```

# **Stage 4 : Key Findings, Business Insights & Recommendations**

### **Key Findings**



*   Overall GDP Growth: GDP grew from  1972 to  2026, giving 40x growth in 54 years.
*   Production Dominance: Production measure has highest records, covering all industries from Agriculture to Services.

*   Service Sector: Top industries in 2026 are Professional Services, Health, Rental & Real Estate. Services are backbone of NZ economy.
*   Expenditure: Household Consumption is biggest driver, followed by Investment and Exports.

*   Seasonal Pattern: December and June quarters show higher GDP, March shows lower.
*   COVID Impact: Sharp fall in June 2020 quarter, strong recovery in 2021.

*   Decade Growth: Average GDP in 2020s is highest, showing continuous expansion despite shocks.

*   Correlation: Year and GDP have strong positive correlation (0.88), showing long-term growth trend.










### **Business Insights**



*   Service sector is associated with GDP growth. Focus on Professional Services, Health and Technology to drive future growth.
*   Household Consumption is a large, stable driver. Analyze where consumption drops and whether price, confidence or employment creates friction.

*   Returning to growth after shocks: Economy recovered quickly after COVID. Examine stimulus, export resilience and domestic consumption paths.

*   Seasonal differences deserve validation. December is strongest, but campaign, tourism and harvest factors are not included as explanatory evidence.
*   Expenditure balance: Monitor Exports vs Imports balance. Trade deficit can affect GDP growth.


*  Use correlation carefully. Strong correlation between Year and GDP reflects long-term trend, not causation.





### **Recommendations**


*   Improve service sector productivity: Test digital adoption, skill training, and technology investment in high-growth services.

*  Investigate consumption journey: Build funnel from income to spending to saving. Identify drop-off in consumption.

*  Run investment promotion: Promote investment in high-growth industries, infrastructure and green energy.
*   Audit lower-contributing industries: Review Agriculture and Manufacturing productivity, export competitiveness and technology use.

*   Validate seasonal patterns: Compare quarterly results with tourism, harvest and campaign data. Run controlled policy tests before changing spending.

*   Monitor consistent KPI set: Track GDP, GDP per capita, industry growth, consumption, investment and trade balance over time
*  Collect richer context: Add employment, inflation, interest rate and global trade data to support more specific recommendations.
